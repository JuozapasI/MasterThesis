## Loading data

```python
samples = [
    "PBMC_10x", "PBMC_10x_2", "PBMC_10x_3", "PBMC_indrops", "PBMC_indrops_2", 
    "brain", "brain_2", "eye", "eye_2", "eye_3", "lung_2", "lung_5", "lung_7", "lung_8"
]
```

```python
adatas = {}

for sample in samples:
    adatas[sample] = sc.read_10x_mtx(f'../../data/downstream/matrices/intergenic/{sample}/')
    adatas[sample].raw = adatas[sample].copy()
```

# Normalizing data

```python
for sample in samples:
    sc.pp.normalize_total(adatas[sample], target_sum=None)
    sc.pp.log1p(adatas[sample])
```

# Save matrices

```python
for sample in samples:
    adatas[sample].write(f"../../data/downstream/adatas/intergenic_combined/{sample}.h5ad")
```
